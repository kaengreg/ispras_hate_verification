```python
import pandas as pd
```

```python
RANDOM_STATE = 13
fraction = 1

input_file_path_1 = "/Users/kaengreg/Documents/Работа /НИВЦ/ispras/data/countries_classification/train_raw.jsonl"
input_file_path_2 = "/Users/kaengreg/Documents/Работа /НИВЦ/ispras/data/toxicity_classification/ru_social_tox_stance_26_04_train.jsonl"
output_file_path = "/Users/kaengreg/Documents/Работа /НИВЦ/ispras/data/toxicity_classification/train_mixed_stance_tox.jsonl"
```

```python
def data_read(path: str) -> pd.DataFrame:
    return pd.read_json(path, lines=True)

train_data = data_read(input_file_path_1)
train_data
```

```python
refined_train = train_data.sample(frac=fraction, random_state=RANDOM_STATE)
refined_train
```

```python
train_data_addition = data_read(input_file_path_2)
train_data_addition
```

```python
refined_train = pd.concat([refined_train, train_data_addition], ignore_index=True)
refined_train
```

```python
refined_train = refined_train.sample(frac=1, random_state=RANDOM_STATE).reset_index(drop=True)
refined_train = refined_train.fillna(value={"toxicity_label":"UNDEF"})
refined_train
```

```python
refined_train.to_json(output_file_path, orient="records", lines=True, force_ascii=False)
print(f"Saved {len(refined_train)} rows to {output_file_path}")
```
